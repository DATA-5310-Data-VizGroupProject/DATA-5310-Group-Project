# Healthcare coverage

## Data prep

```python
import pandas as pd
import altair as alt
```

```python
df = pd.read_csv('/Users/vincent/Downloads/OECD.ELS.HD,DSD_HEALTH_PROT@DF_HEALTH_PROT,1.1,filtered,2026-11-14 19-39-11.csv')
```

```python
df.head()
```

```python
keep = ['Reference area', 'TIME_PERIOD', 'INSURANCE_TYPE', 'Insurance type', 'UNIT_MEASURE', 'OBS_VALUE']

df = df[keep]
df.columns = ['Country', 'Year', 'Code', 'Insurance Type', 'Unit', 'OBS_Value']
```

```python
df.head()
```

```python
df = df[df['Unit'] == 'PT_POP'].copy()
```

```python
df= df[df['Country'].isin(['United States', 'Canada'])].copy()
```

```python
df = df[df['Year'] <= 2024]
```

```python
df= df.sort_values(['Country', 'Code', 'Year']).reset_index(drop=True)
```

```python
df.head(20)
```

```python
cover = df[df['Code'] == 'TPRIBASI'].copy()
cover['Insurance Type'] = 'Uninsured (estimated)'
cover['Code'] = 'UNINSURED'
cover['OBS_Value'] = 100 - cover['OBS_Value']
df = pd.concat([df, cover], ignore_index=True)
```

```python
alt.Chart(df).mark_line().transform_filter(
    alt.datum.Code == 'TPRIBASI'
).encode(
    alt.X('Year:O'),
    alt.Y('OBS_Value:Q'),
    alt.Color('Country:N', scale=alt.Scale(scheme='viridis'), title='Country')
)
```

```python
alt.Chart(df).mark_line().transform_filter(
    (alt.datum.Country == 'United States') &
    ((alt.datum.Code == 'COVGCMED') | (alt.datum.Code == 'PHINPMPI') | (alt.datum.Code == 'PHINCMPI'))
).encode(
    alt.X('Year:O'),
    alt.Y('OBS_Value:Q'),
    alt.Color('Insurance Type:N', scale=alt.Scale(scheme='viridis'), title='Insurance Type')
)
```

```python
alt.Chart(df).mark_bar().transform_filter(
    (alt.datum.Country == 'United States') & (alt.datum.Year == 2024) &
    (alt.datum.Code != 'TPRIBASI') & (alt.datum.Code != 'UNINSURED')
).encode(
    alt.X('OBS_Value:Q'),
    alt.Y('Insurance Type:N', sort='-x')
)
```

```python
alt.Chart(df).mark_line().transform_filter(
    alt.datum.Code == 'UNINSURED'
).encode(
    alt.X('Year:O'),
    alt.Y('OBS_Value:Q'),
    alt.Color('Country:N', scale=alt.Scale(scheme='viridis'), title='Country')
)
```
